```python
X = df.drop('price', axis=1)
y = df['price']
```

```python
X
```

```python
X_one_encode = pd.get_dummies(X,columns=['model', 'transmission', 'fuelType'], drop_first=True)
```

```python
X_one_encode = X_one_encode.astype(int)
```

```python
X_one_encode
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

columns = ['model', 'transmission', 'fuelType']
Xlabel = X
for col in columns:
    Xlabel[col] = le.fit_transform(Xlabel[col])

```

```python
Xlabel
```

```python
#Now we will check which performs better 1 encode or label encode
from sklearn.preprocessing import StandardScaler
```

```python
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
scaler = StandardScaler()

X_one_encode[numerical_cols] = scaler.fit_transform(X_one_encode[numerical_cols])
X_one_encode
```

```python
Xlabel.columns
```

```python
Xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scaler.fit_transform(Xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])
```

```python
Xlabel
```

```python
#model with one encoding
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_one_encode, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
y_test
```

```python
r2 = r2_score(y_test, y_pred)
r2
```

```python
#adjusted r2
n = X_test.shape[0] # Number of observations (rows)
p = X_test.shape[1] # Number of features (columns)
adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
adj_r2
```
